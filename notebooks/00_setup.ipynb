## Step 3: Sanity check

Let's make one tiny agent call to confirm everything is working before we go any further.


```python
from agents import Agent, Runner

agent = Agent(name="Sanity Check Agent", instructions="Reply with just the word: works")
result = Runner.run_sync(agent, "Are you working?")
print(result.final_output)
```

If that printed something like `works`, you're all set.
