# 02 - Giving Your Agent Tools

An LLM on its own can only *talk*. It doesn't know today's date, can't reliably do exact arithmetic, and can't look anything up. **Tools** are how you give it the ability to actually *do* things.

A tool is just a normal Python function. You decorate it with `@function_tool`, and the agent can now choose to call it.

## The `@function_tool` decorator

Two things matter a lot here:
- **The docstring** -- the agent reads this to understand what the tool does and when to use it.
- **Type hints** -- they tell the agent what kind of arguments to pass in.

A vague docstring means the agent won't know when to reach for the tool, or will use it wrong.


```python
from agents import Agent, Runner, function_tool

@function_tool
def get_current_time() -> str:
    """Return the current date and time."""
    from datetime import datetime

    return datetime.now().strftime("%A, %d %B %Y, %H:%M")
```

Now give the tool to an agent via the `tools=` list:


```python
agent = Agent(
    name="Time-aware Assistant",
    instructions="You are a helpful assistant. Use your tools when they help answer the question.",
    tools=[get_current_time],
)

result = Runner.run_sync(agent, "What time is it right now?")
print(result.final_output)
```

## A tool with an argument

Function parameters become arguments the agent can choose to fill in. Give parameters a docstring `Args:` section so the agent knows what each one means.


```python
@function_tool
def roll_dice(sides: int = 6) -> str:
    """Roll a dice and return the result.

    Args:
        sides: How many sides the dice has.
    """
    import random

    return f"You rolled a {random.randint(1, sides)} (out of {sides})."


agent = Agent(
    name="Game Master",
    instructions="You are a game master running a tabletop game. Use your tools when needed.",
    tools=[get_current_time, roll_dice],
)

result = Runner.run_sync(agent, "Roll a 20-sided dice for my attack.")
print(result.final_output)
```
