# 01 - Your First Agent

At its core, an **agent** is just two things:
1. **Instructions** -- a description of who it is and how it should behave (this is often called a "system prompt").
2. **A model** -- the LLM that actually generates responses.

Everything else you'll learn this course (tools, memory, hosted capabilities) is built on top of that simple idea.

## Creating an agent

`Agent` takes a `name` (for your own bookkeeping/logs -- the user never sees it) and `instructions` (the personality + rulebook).


```python
from agents import Agent, Runner

agent = Agent(
    name="History Tutor",
    instructions="You answer history questions clearly and concisely, in two sentences or less.",
)
```

## Running it

`Runner.run_sync(agent, message)` sends a message to the agent and waits for the full response. The result object has a `.final_output` -- the agent's actual reply.


```python
result = Runner.run_sync(agent, "When did the Roman Empire fall?")
print(result.final_output)
```

## A note on sync vs. async

We used `Runner.run_sync` here because it's the simplest way to call an agent from a notebook or script. There's also an async version, `await Runner.run(agent, message)`, which is what the Telegram bot uses later (web servers need to handle many requests at once, which is what async is for). They work identically otherwise -- everything you learn with `run_sync` carries straight over.
