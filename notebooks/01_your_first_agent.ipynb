## Creating an agent

`Agent` takes a `name` (for your own bookkeeping/logs -- the user never sees it) and `instructions` (the personality + rulebook).


```python
from agents import Agent, Runner

agent = Agent(
    name="History Tutor",
    instructions="You answer history questions clearly and concisely, in two sentences or less.",
)
```

## Running it

`await Runner.run(agent, message)` sends a message to the agent and waits for the full response. The result object has a `.final_output` -- the agent's actual reply.

The `await` is there because Jupyter notebook cells support "top-level await" -- you can await a coroutine directly in a cell, no extra setup needed. (There's also a `Runner.run_sync(...)` that skips the `await` for plain scripts -- see the note at the end of this notebook for why we don't use it here.)


```python
result = await Runner.run(agent, "When did the Roman Empire fall?")
print(result.final_output)
```

## A note on `await`

The SDK actually has two ways to run an agent: `await Runner.run(...)` (async) and `Runner.run_sync(...)` (a synchronous wrapper around it, for plain scripts). You might expect the notebook-friendly one to be `run_sync` -- it has "sync" right there in the name, and no `await` to think about. It's actually the opposite: `run_sync` **does not work in Jupyter notebooks** (the SDK's own docs say so explicitly), because Jupyter cells already have an event loop running in the background, and `run_sync` tries to start its own -- Python doesn't allow starting one event loop inside another.

`await Runner.run(...)` avoids the problem entirely: instead of starting a new event loop, it just uses the one Jupyter already has running. That's why every notebook in this course uses `await Runner.run(...)`. `Runner.run_sync(...)` is what you'd reach for in a plain script (`python my_script.py`) with no event loop already running -- not here.
