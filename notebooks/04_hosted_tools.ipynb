## Web search

`WebSearchTool()` gives your agent access to live web search -- useful for anything after the model's training cutoff, or anything that changes day to day.


```python
from agents import Agent, Runner, WebSearchTool

agent = Agent(
    name="Research Assistant",
    instructions="Use web search for anything current or time-sensitive.",
    tools=[WebSearchTool()],
)

result = Runner.run_sync(agent, "What's a major tech news story from today?")
print(result.final_output)
```

## Code interpreter

`CodeInterpreterTool()` gives your agent a real, sandboxed Python environment (with pandas etc. available). This matters because LLMs are unreliable at exact arithmetic on their own -- for anything that needs to be precisely *computed*, this is the tool you want, not the model's own "mental math".


```python
from agents import CodeInterpreterTool

agent = Agent(
    name="Data Analyst",
    instructions="Use code_interpreter for anything that needs precise computation.",
    tools=[
        CodeInterpreterTool(tool_config={"type": "code_interpreter", "container": {"type": "auto"}}),
    ],
)

result = Runner.run_sync(
    agent,
    "What is the compound annual growth rate if $1000 grows to $2500 over 6 years? Show your work.",
)
print(result.final_output)
```

Code interpreter can also *create files* -- e.g. build an actual spreadsheet from data, not just describe it in text. Your capstone Telegram bot uses exactly this: it fetches data with a custom tool, then asks code interpreter to build a `.xlsx` report from it, and automatically sends you the resulting file.
