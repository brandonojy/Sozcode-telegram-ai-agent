One difference you'll notice in the repo that isn't a new concept, just plumbing:

- `await Runner.run(...)` becomes `await Runner.run_streamed(...)` -- same idea, but it hands you events (tool calls, reasoning, the final message) one at a time as they happen, instead of making you wait for everything at once. That's what lets `app.py` send the trace messages live rather than all at once at the end.
- The agent doesn't actually run inside the webhook request itself. `app.py` is split into `/webhook` (replies to Telegram immediately) and `/process` (does the real work moments later) -- purely so Telegram never has to wait around for a slow response. It doesn't change anything about the agent code itself.

## If you finish early: bonus concepts

These are beyond what this course covers, but worth knowing exist if you want to keep exploring after today:

- **Handoffs** -- one agent can hand a conversation off to a different specialist agent (e.g. a triage agent routing to a billing agent or a tech-support agent). See [Handoffs](https://openai.github.io/openai-agents-python/handoffs/) in the docs.
- **Structured outputs** -- instead of a plain string reply, you can make an agent return data matching a schema you define (e.g. a Pydantic model), which is useful when another program needs to consume the result. See [Structured outputs](https://openai.github.io/openai-agents-python/agents/#structured-outputs) in the docs.
- **Guardrails** -- checks that run before/after the agent, to validate input or catch bad output. See [Guardrails](https://openai.github.io/openai-agents-python/guardrails/) in the docs.
